```python
df_model=df.drop(columns=['msisdn','pcircle','pdate'])
```

```python
df_model.columns.tolist()
```

```python
x=df_model.drop('label',axis=1)
y=df_model['label']
```

```python
print("X shape:",x.shape)
print("Y shape:",y.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y)
```

```python
print("X_train:",x_train.shape)
print("x_test:",x_test.shape)
print("y_train:",y_train.shape)
print("y_test:",y_test.shape)
```

```python
print("Training target distribution:")
print(y_train.value_counts(normalize=True)*100)
print("Testing target distribution:")
print(y_test.value_counts(normalize=True)*100)

```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
log_model=LogisticRegression(
    class_weight='balanced',
    max_iter=1000,
    random_state=42
)
log_model.fit(x_train_scaled,y_train)
```

```python
y_pred_log=log_model.predict(x_test_scaled)
y_prob_log=log_model.predict_proba(x_test_scaled)[:,1]
```

```python
from sklearn.metrics import log_loss, precision_score,recall_score
log_loss_value=log_loss(y_test,y_prob_log)
precision=precision_score(y_test,y_pred_log)
recall=recall_score(y_test,y_pred_log)
print("Log loss:",log_loss_value)
print("Precision:",precision)
print("Recall:",recall)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt_model=DecisionTreeClassifier(
    random_state=42,
    class_weight='balanced'
)
dt_model.fit(x_train,y_train)
```

```python
y_pred_dt=dt_model.predict(x_test)
y_prob_dt=dt_model.predict_proba(x_test)[:,1]
```

```python
log_loss_dt=log_loss(y_test,y_prob_dt)
precision_dt=precision_score(y_test,y_pred_dt)
recall_dt=recall_score(y_test,y_pred_dt)
print("Decision Tree Log Loss:",log_loss_dt)
print("Decision Tree Precision:",precision_dt)
print("Decision Tree Recall:",recall_dt)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model=RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight='balanced',
    n_jobs=-1
)
rf_model.fit(x_train,y_train)
```

```python
y_pred_rf=rf_model.predict(x_test)
y_prob_rf=rf_model.predict_proba(x_test)[:,1]
```

```python
log_loss_rf=log_loss(y_test,y_prob_rf)
precision_rf=precision_score(y_test,y_pred_rf)
recall_rf=recall_score(y_test,y_pred_rf)
print("Random Forest Log Loss:",log_loss_rf)
print("Random forest Precision:",precision_rf)
print("Random Forest Recall:",recall_rf)
```

```python
from sklearn.ensemble import GradientBoostingClassifier
gb_model=GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)
gb_model.fit(x_train,y_train)
```

```python
y_pred_gb=gb_model.predict(x_test)
y_prob_gb=gb_model.predict_proba(x_test)[:,1]
```

```python
log_loss_gb=log_loss(y_test,y_prob_gb)
precision_gb=precision_score(y_test,y_pred_gb)
recall_gb=recall_score(y_test,y_pred_gb)
print("Gradient Boosting Log Loss:",log_loss_gb)
print("Gradient Boosting Precision:",precision_gb)
print("Gradient Boosting Recall:",recall_gb)
```

```python
from sklearn.naive_bayes import GaussianNB
nb_model=GaussianNB()
nb_model.fit(x_train_scaled,y_train)
```

```python
y_pred_nb=nb_model.predict(x_test_scaled)
y_prob_nb=nb_model.predict_proba(x_test_scaled)[:,1]
```

```python
log_loss_nb=log_loss(y_test,y_prob_nb)
precision_nb=precision_score(y_test,y_pred_nb)
recall_nb=recall_score(y_test,y_pred_nb)
print("Naive Bayes Log Loss:",log_loss_nb)
print("Naive Bayes Precision:",precision_nb)
print("Naive Bayes Recall:",recall_nb)
```

```python
results=pd.DataFrame({
    'Model':[
        'Logistic Regression',
        'Decision Tree',
        'Random Forest',
        'Gradient Boosting',
        'Gaussian Naive Bayes'
    ],
    'Log Loss':[
        log_loss_value,
        log_loss_dt,
        log_loss_rf,
        log_loss_gb,
        log_loss_nb
    ],
    'Precision':[
        precision,
        precision_dt,
        precision_rf,
        precision_gb,
        precision_nb
    ],
    'Recall':[
        recall,
        recall_dt,
        recall_rf,
        recall_gb,
        recall_nb]

        })
        
```

```python
results
```

```python
results.sort_values(by='Log Loss')
```

```python
best_model_row=results.loc[results['Log Loss'].idxmin()]
print("Best Model:",best_model_row['Model'])
print("Log loss:",best_model_row['Log Loss'])
print("Precision:",best_model_row['Precision'])
print("Recall:",best_model_row['Recall'])
```

```python
from sklearn.model_selection import RandomizedSearchCV
param_grid={
    'n_estimators':[50,100],
    'learning_rate':[0.05,0.1],
    'max_depth':[2,3]
}
```

```python
gb_search= RandomizedSearchCV(
     GradientBoostingClassifier(random_state=42),
     param_distributions=param_grid,
     n_iter=2,
    scoring='neg_log_loss',
    cv=2,
    random_state=42,
    n_jobs=-1
)
```

```python
gb_search.fit(x_train,y_train)
```

```python
print("Best Performance:")
print(gb_search.best_params_)
print("\nBest CV Score:")
print(gb_search.best_score_)
```

```python
best_gb=gb_search.best_estimator_
```

```python
y_pred_tuned=best_gb.predict(x_test)
y_prob_tuned=best_gb.predict_proba(x_test)[:,1]
```

```python
tuned_log_loss=log_loss(y_test,y_prob_tuned)
tuned_precision=precision_score(y_test,y_pred_tuned)
tuned_recall=recall_score(y_test,y_pred_tuned)
```

```python
print("Tuned Gradient Boosting Performance:")
print("Log loss:",tuned_log_loss)
print("Precision:",tuned_precision)
print("Recall:",tuned_recall)
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt
auc_lr=roc_auc_score(y_test,y_prob_log)
auc_dt=roc_auc_score(y_test,y_prob_dt)
auc_rf=roc_auc_score(y_test,y_prob_rf)
auc_gb=roc_auc_score(y_test,y_prob_gb)
auc_nb=roc_auc_score(y_test,y_prob_nb)
auc_tuned_gb=roc_auc_score(y_test,y_prob_tuned)
print("Logistic Regression AUC:",auc_lr)
print("Decision Tree AUC:",auc_dt)
print("Random Forest AUC:",auc_rf)
print("Gradient Boosting AUC:",auc_gb)
print("Gaussian Naive Bayes AUC:",auc_nb)
print("Tuned Gradient Boosting  AUC:",auc_tuned_gb)
```

```python
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_prob_log)
fpr_dt, tpr_dt, _ = roc_curve(y_test, y_prob_dt)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_prob_rf)
fpr_gb, tpr_gb, _ = roc_curve(y_test, y_prob_gb)
fpr_nb, tpr_nb, _ = roc_curve(y_test, y_prob_nb)
fpr_tuned_gb, tpr_tuned_gb, _ = roc_curve(y_test, y_prob_tuned)

plt.figure(figsize=(10, 7))

plt.plot(fpr_lr, tpr_lr, label=f"Logistic Regression (AUC = {auc_lr:.3f})")
plt.plot(fpr_dt, tpr_dt, label=f"Decision Tree (AUC = {auc_dt:.3f})")
plt.plot(fpr_rf, tpr_rf, label=f"Random Forest (AUC = {auc_rf:.3f})")
plt.plot(fpr_gb, tpr_gb, label=f"Gradient Boosting (AUC = {auc_gb:.3f})")
plt.plot(fpr_nb, tpr_nb, label=f"Gaussian Naive Bayes (AUC = {auc_nb:.3f})")
plt.plot(fpr_tuned_gb, tpr_tuned_gb,
         label=f"Tuned Gradient Boosting (AUC = {auc_tuned_gb:.3f})")

plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Model Comparison")
plt.legend()
plt.grid()
plt.show()
```
